# Natural Language Processing (NLP) Analysis

NLP is used here to analyse the text in SIH 2026 problem statement titles.

The main purpose is to find the important words and topics that appear frequently in the problem statements. This helps us understand the major technology areas and requirements mentioned in the challenges.

In this analysis, we will:
- Clean the problem statement titles
- Remove common words
- Find frequently used words
- Identify important technology-related terms
- Analyse the main topics in the problem statements
- Use the results later for the recommendation system

```python
from collections import Counter
import re

text = " ".join(df["Problem Statement Title"].astype(str)).lower()

words = re.findall(r"\b[a-z]+\b", text)

word_counts = Counter(words)

word_counts.most_common(100)
```

```python
stop_words = {
    "and", "for", "of", "in", "to", "a", "an", "the",
    "on", "with", "by", "from", "using", "that", "through",
    "based", "into", "as", "is", "are", "be", "or", "this"
}

important_words = [
    word for word in words
    if word not in stop_words
]

important_counts = Counter(important_words)

important_counts.most_common(20)
```

## TF-IDF Analysis

### Question

Which words are important in each SIH 2026 problem statement?

TF-IDF gives more importance to words that are useful for a particular problem statement and less importance to words that appear in many problem statements.

This will help us compare problem statements based on their text and will be used later in the recommendation system.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform(
    df["Problem Statement Title"])

print("Number of problem statements:", tfidf_matrix.shape[0])
print("Number of important words:", tfidf_matrix.shape[1], "\n")



# Get the words/features identified by TF-IDF
feature_names = vectorizer.get_feature_names_out()

# Show the first 20 words
print(feature_names[:100])
```

```python
# Get the TF-IDF values for the first problem statement
first_statement = tfidf_matrix[0].toarray()[0]

# Show words that have a TF-IDF value greater than zero
for word, value in zip(feature_names, first_statement):
    if value > 0:
        print(word, round(value, 3))
```

## Finding Similar Problem Statements

### Question

Which SIH 2026 problem statements are most similar to each other?

TF-IDF has converted each problem statement into numerical values.

Cosine similarity will compare these values and give a similarity score between problem statements.

A higher score means the problem statements are more similar in terms of their words and content.

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(tfidf_matrix)

print("Similarity matrix shape:", similarity_matrix.shape)
```

## Finding Similar Problem Statements

### Question

Which problem statements are most similar to the first SIH 2026 problem statement?

We will compare the first problem statement with all other problem statements using cosine similarity.

```python
# Get similarity scores for the first problem statement
similarity_scores = similarity_matrix[0]

# Find the 5 most similar problem statements
similar_indices = similarity_scores.argsort()[-6:][::-1]

# Display the results
for index in similar_indices:
    print(index, round(similarity_scores[index], 3))
```

```python
similarity_scores = similarity_matrix[0]

print(similarity_scores)
```

Finding the Most Similar Problem Statements
Question   

Which SIH 2026 problem statements are most similar to the first problem statement?

We already calculated the similarity between all 226 problem statements. Now we will find the 5 highest similarity scores for the first problem statement

```python
# Get similarity scores for the first problem statement
similarity_scores = similarity_matrix[0].copy()

# Remove the first problem itself
similarity_scores[0] = -1

# Find the 5 most similar problem statements
top_indices = similarity_scores.argsort()[-5:][::-1]

# Display the results
for index in top_indices:
    print("Similarity:", round(similarity_scores[index], 3))
    print("Problem:", df.loc[index, "Problem Statement Title"])
    print()
```

## SIH 2026 Problem Statement Recommendation

### Question

Can we recommend relevant SIH 2026 problem statements based on a user's requirement?

The TF-IDF model can convert a user's requirement into the same numerical representation used for the problem statements.

Cosine similarity is then used to find the problem statements that are most similar to the user's requirement.

This forms the basic recommendation system for the project.

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_problems(user_query, top_n=5):
    
    # Convert the user's query into TF-IDF
    query_vector = vectorizer.transform([user_query])
    
    # Compare the query with all problem statements
    similarity_scores = cosine_similarity(
        query_vector,
        tfidf_matrix
    )[0]
    
    # Get the indexes of the most similar problems
    top_indices = similarity_scores.argsort()[-top_n:][::-1]
    
    # Display recommendations
    for index in top_indices:
        print("Similarity:", round(similarity_scores[index], 3))
        print("Problem:", df.loc[index, "Problem Statement Title"])
        print("PS Number:", df.loc[index, "PS Number"])
        print("Organization:", df.loc[index, "Organization"])
        print("Theme:", df.loc[index, "Theme"])
        print()
```

## Testing the Recommendation System

The recommendation system can now accept a user's requirement and find the most similar SIH 2026 problem statements.

For testing, we will provide a sample requirement related to disaster management and early warning systems.

```python
user_query = "AI based disaster early warning and monitoring system"

recommend_problems(user_query)
```

## Recommendation Results

The recommendation system returns the problem statements that are most similar to the user's requirement.

The results include the similarity score along with the problem statement details, making it easier to compare and identify relevant challenges.

```python
# Get recommendations for a sample user requirement

user_query = "AI based disaster early warning and monitoring system"

query_vector = vectorizer.transform([user_query])

similarity_scores = cosine_similarity(
    query_vector,
    tfidf_matrix
)[0]

top_indices = similarity_scores.argsort()[-5:][::-1]

recommendations = df.loc[
    top_indices,
    ["PS Number", "Problem Statement Title", "Organization", "Theme"]
].copy()

recommendations.insert(
    0,
    "Rank",
    range(1, len(recommendations) + 1)
)

recommendations["Similarity Score"] = [
    round(similarity_scores[index], 3)
    for index in top_indices
]

recommendations
```

## Testing Different User Requirements

The recommendation system should work with different types of requirements, not only disaster management.

We will test it with a few different technology-related requirements and check whether the recommended problem statements are relevant.

```python
test_queries = [
    "cybersecurity threat detection system",
    "smart agriculture system for farmers",
    "AI based healthcare monitoring system"
]

for query in test_queries:
    print("\nUser Requirement:", query)
    print("-" * 60)
    recommend_problems(query, top_n=3)
```

## Recommendation System Testing

The recommendation system was tested using different user requirements related to cybersecurity, agriculture and healthcare.

The system produced relevant recommendations for many queries, especially when the query contained terms that were closely related to the problem statement titles.

However, some recommendations were less relevant, particularly for broader queries such as healthcare monitoring. This is because the current model uses TF-IDF and cosine similarity based mainly on text similarity rather than deeper semantic understanding.

This shows that the current recommendation system provides a useful baseline, but it can be improved in future versions using more advanced NLP techniques.

 SIH Problem Selection
Identify AI/ML-Friendly Software Problems

The SIH 2026 dataset contains both Software and Hardware problem statements.

Since this project is focused on Data Science, Artificial Intelligence and Machine Learning, we first filter the dataset to identify Software problems related to AI/ML.

Keywords such as AI, Machine Learning, prediction, detection, forecasting, recommendation, analytics and monitoring are used to create an initial shortlist.

This is only a preliminary filtering step. The shortlisted problems will be evaluated further based on their difficulty, data requirements, AI/ML suitability and feasibility for implementation.

```python
# Shortlist AI/ML related problems

keywords = [
    "AI", "AI-based", "AI-enabled", "AI-powered",
    "machine learning", "ML",
    "prediction", "predictive",
    "detection", "classification",
    "forecast", "forecasting",
    "recommendation",
    "analytics", "monitoring",
    "recognition", "intelligent"
]

software_problems = df[df["Category"].str.lower() == "software"].copy()

pattern = "|".join(keywords)

ai_ml_problems = software_problems[
    software_problems["Problem Statement Title"]
    .str.contains(pattern, case=False, na=False)
]

ai_ml_problems = ai_ml_problems[
    [
        "PS Number",
        "Problem Statement Title",
        "Organization",
        "Theme",
        "Category"
    ]
]

print("AI/ML-friendly Software Problems:", len(ai_ml_problems))

ai_ml_problems
```
